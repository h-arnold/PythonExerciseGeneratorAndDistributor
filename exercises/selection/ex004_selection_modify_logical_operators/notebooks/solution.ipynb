## Goal
Welcome back to **selection**! You already know how to compare two values with `<`, `>`, `<=`, `>=`, `==` and `!=`. In this set you will combine comparisons with the logical operators `and`, `or` and `not`.

You will:
- Add `and` to require two things at once
- Add `or` to accept either of two things
- Swap `and` and `or` to change what counts as a pass
- Use `not` to flip a check around
- Then bring back constants, f-strings and arithmetic (`+`, `*`, `//`) with your new operators

The first four exercises focus only on the operators themselves. Do not worry about constants yet — those return from exercise 5 onwards.

## New ideas

| Operator | Meaning | True example |
|----------|---------|--------------|
| `and` | True when both sides are True | `age >= 11 and age <= 16` is True when age is 14 |
| `or` | True when at least one side is True | `day == "Saturday" or day == "Sunday"` is True when day is Sunday |
| `not` | Flips True to False and False to True | `not (score >= 50)` is True when score is 30 |

Use brackets when you mix `and` with `or`, for example `member == "yes" and (spend >= 40 or age >= 60)`. The brackets tell Python which part to check first.

## Exercise 1 — Add AND to check both ends of the age range

**Working code:**
```python

    print(f"In range: age {age} can join")
else:
    print(f"Out of range: age {age} cannot join")
```

**Scenario:** The after-school club is open to pupils aged 11 to 16. The program below only checks the lower end of the range, so older teenagers slip through when they should not.

**Task:**
- Change the branch so someone is only in range when two checks both pass: the age is 11 or older, and the age is 16 or younger.
- Rewrite the first message so it says the age fits the club, showing the actual age.
- Rewrite the second message so it says the age does not fit, showing the actual age.
- Do not use constants in this exercise.

**Expected output:**
```

```


````python
age_input = input("Enter your age: ")
age = int(age_input)
if age >= 11 and age <= 16:
    print(f"In range: age {age} fits the club")
else:
    print(f"Out of range: age {age} does not fit")
````


**Scenario:** This program splits scores at the 50 boundary. It already works, but the department writes its checks the other way round, so yours must use the word not instead.

## Exercise 10 — Film points tiers (hardest!)

**Working code:**
```python
films_input = input("How many films this month? ")
films = int(films_input)
age_input = input("Enter your age: ")
age = int(age_input)
total = films * 5
BIG_SPEND = 18
STANDARD_SPEND = 14
GOLD_AGE = 12
GOLD_FILMS = 5
if total >= BIG_SPEND:
    print(f"Gold member: £{total} for age {age}")
elif total >= STANDARD_SPEND:
    print(f"Silver member: £{total} for age {age}")
else:
    print(f"Bronze member: £{total} for age {age}")
```

**Scenario:** The cinema loyalty scheme has gold, silver and bronze tiers based on points, with a small tenth-off adjustment. The starter awards each tier on points alone, but each tier has an extra eligibility rule to add: gold needs a high total plus either the gold age or enough films, while silver needs a standard total plus a minimum age.

**Task:** This is the hardest one, in three parts.
- First, add the discount calculation: directly after the line that works out the total from the films, add a line that finds a tenth of that total (rounded down), then a line that takes that amount off the total.
- Next, change the gold branch so it needs a high total plus either reaching the gold age or having watched enough films.
- Then change the silver branch so it needs a standard total while aged 5 or over.
- Keep all four constants and all three messages exactly as they are.

**Expected output:**
```
How many films this month? [⌨️: 6]
Enter your age: [⌨️: 20]
Gold member: £27 for age 20
```

````python
films_input = input("How many films this month? ")
films = int(films_input)
age_input = input("Enter your age: ")
age = int(age_input)
total = films * 5
discount = total // 10
total = total - discount
BIG_SPEND = 18
STANDARD_SPEND = 14
GOLD_AGE = 12
GOLD_FILMS = 5
if total >= BIG_SPEND and (age >= GOLD_AGE or films >= GOLD_FILMS):
    print(f"Gold member: £{total} for age {age}")
elif total >= STANDARD_SPEND and not (age < 5):
    print(f"Silver member: £{total} for age {age}")
else:
    print(f"Bronze member: £{total} for age {age}")
````
